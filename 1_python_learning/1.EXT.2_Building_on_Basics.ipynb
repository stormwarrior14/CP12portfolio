## Quick Recall: Loops

`for i in range(n):` repeats a fixed number of times; `while condition:` repeats until the condition goes `False`. `break` exits a loop early.

```python
# Example
for room_number in range(3):
    print(f"Exploring room {room_number + 1}...")
```

**Quick check:** write a `while` loop that reduces a monster's health by a fixed amount each pass, printing the health each time, until it's at or below 0.

```python
# Your quick check here
import time
health = 100
while health > 0:
    health = health - 10
    print("The monster's health is " + str(health))
    time.sleep(1)
```

## Quick Recall: Functions

`def name(params):` defines a function; `return` sends a value back (no `return` gives `None`). A variable created inside a function stays inside it (scope).

```python
# Example
def calculate_total(gold, bonus):
    return gold + bonus

print(calculate_total(40, 15))
```

**Quick check:** write a function `can_afford(gold, price)` that returns a `bool`, and call it twice with different values.

```python
# Your quick check here
def can_afford(gold,price):
    if price > gold:
        print(False)
    elif price <= gold:
        print(True)
print(can_afford(30, 40))
```

## Quick Recall: Lists &amp; Strings

Lists (`["sword", "shield"]`) are ordered, indexed from `0`, mutable — `.append()` adds, `len()` counts. Strings support the same indexing/slicing but are immutable.

```python
# Example
inventory = ["sword", "shield"]
inventory.append("potion")
print(len(inventory), inventory[0])
```

**Quick check:** given `party = ["Zara", "Old Finn"]`, add a third member and print the full list reversed (`[::-1]`).

```python
# Your quick check here
party = ["Zara", "Old Finn"]
party.append("Birdman")
print(party[::-1])
```

## Quick Recall: Debugging

Read tracebacks bottom-up — the last line names the error and usually describes it clearly. `NameError`, `TypeError`, and `IndexError` each point to a different category of problem. When nothing crashes but the output's still wrong, add a `print()` to see what a variable actually holds.

```python
# Example — a fixed bug, for reference
inventory = ["sword", "shield", "potion"]
print(inventory[len(inventory) - 1])   # last item, without hardcoding the index
```
